```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("products.csv")

# Display first 5 records
print("First 5 Records:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

# Clean product category
df["product category"] = (
    df["product category"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Convert price into numeric values
df["price"] = pd.to_numeric(
    df["price"], errors="coerce"
)

# Convert customer rating into numeric values
df["customer rating"] = pd.to_numeric(
    df["customer rating"], errors="coerce"
)

# Fill missing values
df["product category"] = df["product category"].fillna("Unknown")

df["product name"] = df["product name"].fillna("Unknown")

df["customer rating"] = df["customer rating"].fillna(
    df["customer rating"].median()
)

# Remove rows with missing prices
df.dropna(subset=["price"], inplace=True)

# Price summary statistics
print("\nPrice Summary Statistics:")
print(df["price"].describe())

# Set theme
sns.set_theme(style="whitegrid")

# Create figure
plt.figure(figsize=(10, 6))

# Box plot
sns.boxplot(
    data=df,
    x="product category",
    y="price",
    hue="product category",
    palette="Set2"
)

plt.title("Product Price Analysis by Category")
plt.xlabel("Product Category")
plt.ylabel("Price")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

# Violin plot
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="product category",
    y="price",
    hue="product category",
    palette="Set2"
)

plt.title("Product Price Distribution by Category")
plt.xlabel("Product Category")
plt.ylabel("Price")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
